```python
import numpy as np
rng = np.random.default_rng(seed=42)
OneEvalLogPermutedSamples = rng.choice(
    OneEvalLogSamples,
    size = 10,
    replace=False,
)
```

```python
OneEvalLog.samples
```

```python
OneEvalLogPermutedSamples
```

```python
OneEvalLogPermutedSamplesList = list(OneEvalLogPermutedSamples)
OneEvalLogPermutedSamplesList
```

```python
NewEvalLog = OneEvalLog.copy()
NewEvalLog
```

```python
NewEvalLog.samples = OneEvalLogPermutedSamplesList
```
